# Two Sum

**Problem Statement:** You are given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target. You may assume that each input would have exactly one solution, and you may not use the same element twice.You can return the answer in any order.


**Example 1:**
Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].

**Example 2:**
Input: nums = [3,2,4], target = 6
Output: [1,2]

**Example 3:**
Input: nums = [3,3], target = 6
Output: [0,1]
 

**Constraints:**
2 <= nums.length <= 104
-109 <= nums[i] <= 109
-109 <= target <= 109
Only one valid answer exists.
 

### Approach 1: Brute force (Nested loops)


**Logic**

1. Iterate through the array using an outer loop.
2. For each element, iterate through the elements after it using an inner loop.
3. Check whether their sum equals the target.
4. If a matching pair is found, return their indices.


**Complexity analysis**


**Time:**  O(n²)	Nested loops potentially examine every pair.

**Space:** O(1)	No additional data structure is required.

The main limitation is that as the array grows, the number of comparisons increases significantly.

```python
def two_sum(nums, target):
    for i in range(len(nums) - 1):
        j = i+1
        for j in range(i+1,len(nums)):
            if nums[i] + nums[j] == target:
                return [i,j]

    return 0

nums = [2,8,11,15,7,19]
target = 9
output = two_sum(nums, target)
print(f"Indices: {output}")
```

### Approach 2: Optimized solution using dictionary

The HashMap stores previously visited numbers as keys and their indices as values.

For every current number:
1. Calculate the complement (target - current number).
2. Check whether that complement exists in the HashMap.
3. If it exists, we have found the required pair and can return both indices.
4. Otherwise, store the current number and its index for future iterations.
An important implementation detail: We check the complement before inserting the current number.

This ensures that we never use the same array element twice.

**Complexity analysis**

Complexity	Explanation

**Time:**  O(n) average	We traverse the array once, and HashMap lookups and insertions take O(1) average time.

**Space:**  O(n)	In the worst case, we store almost all elements in the HashMap.

```python
def two_sum_hash(nums, target):
    hashmap = {}

    for i in range(len(nums)):
        diff = target - nums[i]
        if diff in hashmap.keys():
            return [hashmap[diff], i]
        else:
            hashmap[nums[i]] = i


 
nums = [2,8,11,15,7,19]
target = 9

output = two_sum_hash(nums, target)
print(f"Indices: {output}")
```
